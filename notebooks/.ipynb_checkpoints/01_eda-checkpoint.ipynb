# 01 — Exploratory Data Analysis: NYC Taxi Demand

**Goal:** understand hourly pickup demand across NYC's 263 taxi zones before modelling, and justify the feature choices used in production (`src/taxi_demand/features/`).

**Data:** NYC TLC yellow taxi trips, Jan 2023 – Dec 2024, cleaned and aggregated by the pipeline (`make features`) into a zone × hour table.

```python
import json
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from taxi_demand.config import resolve

plt.rcParams.update({"figure.figsize": (11, 3.6), "figure.dpi": 80, "axes.grid": True,
                     "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})

zh = pd.read_parquet(resolve("data/interim/zone_hour.parquet"))
zones = pd.read_csv(resolve("data/raw/taxi_zone_lookup.csv"))
quality = pd.DataFrame(json.loads(resolve("data/interim/quality_report.json").read_text()))

print(f"zone-hour rows : {len(zh):,}")
print(f"zones          : {zh.zone_id.nunique()}")
print(f"hours          : {zh.hour.nunique():,}  ({zh.hour.min():%Y-%m-%d} -> {zh.hour.max():%Y-%m-%d})")
print(f"total pickups  : {zh.pickups.sum():,}")
```

## 1. Data quality
Each month's raw trips are filtered (wrong month, dropoff before pickup, impossible distance/fare, unknown zone). The pipeline **fails** if more than 10% of a month is dropped.

```python
fig, ax = plt.subplots()
ax.bar(quality["month"], 100 * quality["drop_rate"], color="#4C78A8")
ax.axhline(10, color="crimson", ls="--", lw=1, label="quality gate (10%)")
ax.set_ylabel("% trips dropped"); ax.set_title("Rows removed by cleaning, per month")
ax.tick_params(axis="x", rotation=90); ax.legend()
plt.show()
print(f"raw trips: {quality.raw.sum():,} | kept: {quality.kept.sum():,} | "
      f"overall drop rate: {1 - quality.kept.sum() / quality.raw.sum():.2%} | "
      f"worst month: {quality.loc[quality.drop_rate.idxmax(), 'month']} ({quality.drop_rate.max():.2%})")
```

## 2. Demand over time

```python
daily = zh.groupby(zh.hour.dt.floor("D"))["pickups"].sum()
fig, ax = plt.subplots()
ax.plot(daily.index, daily.values, lw=0.6, alpha=0.5, label="daily")
ax.plot(daily.index, daily.rolling(28, center=True).mean(), lw=2, label="28-day mean")
ax.set_title("Citywide pickups per day"); ax.set_ylabel("pickups"); ax.legend()
plt.show()

monthly = zh.groupby(zh.hour.dt.to_period("M"))["pickups"].sum()
yoy = (monthly["2024"].values / monthly["2023"].values - 1) * 100
print("Year-over-year change by month (%):")
print(pd.Series(yoy.round(1), index=[p.strftime("%b") for p in monthly["2024"].index]).to_string())
```

## 3. Daily and weekly seasonality
Average pickups by hour of day and day of week. This is the pattern the calendar features (`hour_of_day`, `day_of_week`, `is_weekend`) and the 24h/168h lags are meant to capture.

```python
city = zh.groupby("hour")["pickups"].sum().to_frame()
city["dow"], city["hod"] = city.index.dayofweek, city.index.hour
grid = city.pivot_table(index="dow", columns="hod", values="pickups", aggfunc="mean")

fig, ax = plt.subplots(figsize=(11, 3.2))
im = ax.imshow(grid.values, aspect="auto", cmap="viridis")
ax.set_yticks(range(7), ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
ax.set_xticks(range(0, 24, 2)); ax.set_xlabel("hour of day"); ax.grid(False)
ax.set_title("Average citywide pickups per hour"); fig.colorbar(im, ax=ax, label="pickups")
plt.show()

peak = grid.stack().idxmax(); low = grid.stack().idxmin()
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
print(f"busiest slot : {days[peak[0]]} {peak[1]:02d}:00  (~{grid.values.max():,.0f} pickups/h)")
print(f"quietest slot: {days[low[0]]} {low[1]:02d}:00  (~{grid.values.min():,.0f} pickups/h)")
```

## 4. Demand is concentrated in a few zones
This matters for evaluation: a metric averaged equally over zones would be dominated by near-empty zones, which is why the project reports **WAPE** (weighted by volume).

```python
by_zone = (zh.groupby("zone_id")["pickups"].sum().sort_values(ascending=False)
           .rename("pickups").reset_index()
           .merge(zones, left_on="zone_id", right_on="LocationID", how="left"))
by_zone["share_%"] = 100 * by_zone.pickups / by_zone.pickups.sum()
print(by_zone.head(10)[["zone_id", "Zone", "Borough", "pickups", "share_%"]]
      .round(2).to_string(index=False))

cum = by_zone["share_%"].cumsum().values
fig, ax = plt.subplots()
ax.plot(np.arange(1, len(cum) + 1), cum)
for k in (10, 50):
    ax.axvline(k, ls=":", color="gray"); ax.text(k + 2, 5, f"top {k}: {cum[k-1]:.0f}%")
ax.set_xlabel("zones ranked by demand"); ax.set_ylabel("cumulative % of pickups")
ax.set_title("Demand concentration across zones")
plt.show()
```

## 5. Sparsity: many zone-hours have zero pickups

```python
zero_share = (zh.pickups == 0).mean()
per_zone_mean = zh.groupby("zone_id")["pickups"].mean()
print(f"zone-hours with zero pickups: {zero_share:.1%}")
print(f"zones averaging < 1 pickup/hour: {(per_zone_mean < 1).sum()} of {len(per_zone_mean)}")
print(f"zones averaging >= 50 pickups/hour: {(per_zone_mean >= 50).sum()}")

fig, ax = plt.subplots()
ax.hist(np.log10(per_zone_mean.clip(lower=0.01)), bins=40, color="#4C78A8")
ax.set_xlabel("log10(mean pickups per hour)"); ax.set_ylabel("zones")
ax.set_title("Zones span 4+ orders of magnitude in demand")
plt.show()
```

## 6. Why lags of 1h, 24h and 168h?
Autocorrelation of the citywide hourly series shows where the past best predicts the present. Peaks at **24h** (same time yesterday) and **168h** (same time last week) confirm the lag features; **1h** captures the current level.

```python
s = city["pickups"]
lags = np.arange(1, 200)
acf = np.array([s.autocorr(l) for l in lags])

fig, ax = plt.subplots()
ax.plot(lags, acf)
for l in (1, 24, 168):
    ax.axvline(l, color="crimson", ls="--", lw=1)
    ax.text(l + 1, acf.min(), f"{l}h: {acf[l-1]:.2f}", color="crimson")
ax.set_xlabel("lag (hours)"); ax.set_ylabel("autocorrelation")
ax.set_title("Autocorrelation of citywide hourly pickups")
plt.show()
```

## 7. Is the data stable over time? (drift preview)
Comparing the average daily profile of 2023 vs 2024. Shifts here are what Phase 6 monitoring will track.

```python
prof = city.assign(year=city.index.year).groupby(["year", "hod"])["pickups"].mean().unstack(0)
fig, ax = plt.subplots()
for y in prof.columns:
    ax.plot(prof.index, prof[y], marker="o", ms=3, label=str(y))
ax.set_xlabel("hour of day"); ax.set_ylabel("avg pickups/hour"); ax.legend()
ax.set_title("Average hourly profile: 2023 vs 2024")
plt.show()
print(f"mean hourly citywide demand: 2023 = {s['2023'].mean():,.0f} | 2024 = {s['2024'].mean():,.0f} "
      f"({(s['2024'].mean() / s['2023'].mean() - 1):+.1%})")
```

## Key takeaways

1. **Clean, consistent data.** 79.5M raw trips → 76.4M kept (3.9% dropped). The worst month (Sep 2023, 5.3%) is still well under the 10% quality gate.
2. **Strong daily and weekly cycles.** Demand ranges from ~190 pickups/hour (Tue 3 am) to ~8,500 (Thu 6 pm), a ~44× swing. Calendar features are essential.
3. **The lag choices are backed by the data.** Autocorrelation of citywide demand: **0.94 at 1h, 0.89 at 24h, 0.92 at 168h**, versus −0.57 at 12h. Same-hour-last-week is nearly as informative as the previous hour, which is why both the `lag_1` and `lag_168` baselines are strong and the model needs both.
4. **Highly concentrated demand.** The top 10 zones (JFK, Upper East Side, Midtown, Penn Station, Times Sq, LaGuardia…) carry ~38% of all pickups, while **176 of 263 zones average under 1 pickup/hour** and **60% of zone-hours are zero**. So evaluation uses **WAPE** (volume-weighted) and a **Poisson** objective (count data with many zeros).
5. **Demand is drifting upward.** 2024 averaged **+7.2%** more pickups/hour than 2023, and the growth accelerated in the autumn (Sep +28.8% YoY, partly reflecting an unusually weak Sep 2023). This is why the model must be **retrained regularly** and why drift monitoring (Phase 6) matters.
